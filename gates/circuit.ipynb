```python
from qiskit import QuantumCircuit
from qiskit import QuantumRegister,ClassicalRegister
```

```python
class Circuit:
    def __init__(self):
        self.qc = QuantumCircuit(2)
        
    def display(self):
        return self.qc.qubits
c = Circuit()
result = c.display()
print(result)
```

## Deutsch's Algorithm

```python
# register with 2 qubits
r1 = QuantumRegister(2,name="qubits")
# register with 1 classical qubit
r2 = ClassicalRegister(1,name="classical")


# combine these:
circuit = QuantumCircuit(r1,r2)
```

```python
circuit.qubits
```

```python
# add some gates
# here our gates.py function will be used
circuit.h(0)
circuit.cx(0,1)

circuit.draw("mpl")
```

```python
qr = QuantumRegister(2, name="qubits")
cr = ClassicalRegister(2, name="classical")
circuit = QuantumCircuit(qr, cr)

circuit.h(qr[0])
circuit.cx(qr[0], qr[1])
circuit.measure(qr, cr)

circuit.draw("mpl")
```

```python
from qiskit.visualization import plot_histogram
from qiskit_aer import AerSimulator 
simulator = AerSimulator()
result = simulator.run(circuit, shots=1000).result()
counts = result.get_counts()

print("Measurement outcomes:", counts)

plot_histogram(counts)
```
